```python
summary = (
    df.groupby("member_casual")
    .agg(
        total_rides=("ride_id", "count"),
        average_duration=("ride_length_minutes", "mean")
    )
    .round(2)
)
summary
```

```python
summary.to_csv("summary_rides.csv")
```
